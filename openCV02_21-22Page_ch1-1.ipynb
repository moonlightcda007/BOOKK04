```python
# iris dataset 불러오기 (공통코드)

import pandas as pd
from sklearn import datasets 

iris = datasets.load_iris()
iris_X = iris.data
iris_y = pd.get_dummies(iris.target).to_numpy()

from sklearn.model_selection import train_test_split 
train_X, test_X, train_y, test_y = train_test_split(iris_X, iris_y, test_size=0.3, random_state=1)
```

```python
# 3.6 Keras를 이용한 인공신경망 분류모델 구현하기 
# Tensorflow 2.x 버전은 케라스를 포함하고 있음. 

# 케라스를 이용한 인공신경망 모델 정의
from tensorflow.keras.models import Sequential 
model = Sequential()

from tensorflow.keras.layers import InputLayer, Dense
model.add(InputLayer(input_shape=(4,))) 
model.add(Dense(50, activation="sigmoid"))
model.add(Dense(30, activation="sigmoid"))
model.add(Dense(3, activation="softmax"))

# 손실함수 정의(Adam), 모델평가 (정확도)
model.compile(loss="categorical_crossentropy", optimizer="adam", metrics=["accuracy"])
```

```python
print("train_X:", train_X.shape)     # 105개 학습 데이터,  4개의 입력 변수
print("train_y:", train_y.shape)     # 3개 클래스 분류   one-hot encoding
print("test_X :", test_X.shape) 
print("test_y :", test_y.shape)   
```

```python
# 모델학습
model.fit(train_X, train_y, epochs=100) 

# 모델평가 
model.evaluate(test_X, test_y) 
```
