```python
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Input, Dense, Dropout, BatchNormalization

model = Sequential([
    Input(shape=(11,)),
    Dense(100, activation='sigmoid'),
    Dropout(0.2),
    BatchNormalization(),
    Dense(200, activation='relu'),
    Dropout(0.4),
    Dense(50, activation='tanh'),
    Dropout(0.1),
    Dense(6, activation='softmax')
])
model.summary()
```

```python
import pandas as pd
import numpy as np

redwine = pd.read_csv('winequality-red.csv', sep=';')
redwine_X = redwine.iloc[:, :-1] # 마지막 열을 제외한 열이 독립변수 X
redwine_y = redwine.iloc[:, -1] - 3 # 마지막 열이 종속변수 y
redwine.head()
```

```python
from sklearn.model_selection import train_test_split
train_X, test_X, train_y, test_y = train_test_split(redwine_X, redwine_y,
                                                    test_size=0.3)
```

```python
model.compile(loss='sparse_categorical_crossentropy',
              optimizer='adam', metrics=['accuracy'])
```

```python
model.fit(train_X, train_y, batch_size=200, epochs=300)
```

```python
model.evaluate(test_X, test_y)
```

```python
import numpy as np
pred = model.predict(test_X)
pred_y = np.argmax(pred, axis=1) + 3
pd.crosstab(test_y+3, pred_y)
```
