```python
# GooLeNet Inception Module 3a  chatGPT 

from tensorflow.keras import Model
from tensorflow.keras.layers import (
    Input,
    Conv2D,
    MaxPooling2D,
    Concatenate
)

# Inception 3a 입력 특징 맵
model_input = Input(
    shape=(28, 28, 192),
    name="input_feature_map"
)

# -------------------------------------------------
# 경로 1: 1×1 Conv, 64
# -------------------------------------------------
branch1 = Conv2D(
    filters=64,
    kernel_size=(1, 1),
    strides=(1, 1),
    padding="same",
    activation="relu",
    kernel_initializer="he_normal",
    use_bias=True,
    name="branch1_1x1"
)(model_input)

# -------------------------------------------------
# 경로 2: 1×1 Conv, 96 → 3×3 Conv, 128
# -------------------------------------------------
branch2 = Conv2D(
    filters=96,
    kernel_size=(1, 1),
    padding="same",
    activation="relu",
    kernel_initializer="he_normal",
    use_bias=True,
    name="branch2_reduce_1x1"
)(model_input)

branch2 = Conv2D(
    filters=128,
    kernel_size=(3, 3),
    padding="same",
    activation="relu",
    kernel_initializer="he_normal",
    use_bias=True,
    name="branch2_3x3"
)(branch2)

# -------------------------------------------------
# 경로 3: 1×1 Conv, 16 → 5×5 Conv, 32
# -------------------------------------------------
branch3 = Conv2D(
    filters=16,
    kernel_size=(1, 1),
    padding="same",
    activation="relu",
    kernel_initializer="he_normal",
    use_bias=True,
    name="branch3_reduce_1x1"
)(model_input)

branch3 = Conv2D(
    filters=32,
    kernel_size=(5, 5),
    padding="same",
    activation="relu",
    kernel_initializer="he_normal",
    use_bias=True,
    name="branch3_5x5"
)(branch3)

# -------------------------------------------------
# 경로 4: 3×3 MaxPool → 1×1 Conv, 32
# -------------------------------------------------
branch4 = MaxPooling2D(
    pool_size=(3, 3),
    strides=(1, 1),
    padding="same",
    name="branch4_3x3_maxpool"
)(model_input)

branch4 = Conv2D(
    filters=32,
    kernel_size=(1, 1),
    padding="same",
    activation="relu",
    kernel_initializer="he_normal",
    use_bias=True,
    name="branch4_pool_projection"
)(branch4)

# -------------------------------------------------
# 채널 방향 결합
# -------------------------------------------------
model_output = Concatenate(
    axis=-1,
    name="inception_3a_concat"
)([branch1, branch2, branch3, branch4])

model = Model(
    inputs=model_input,
    outputs=model_output,
    name="Inception_3a"
)

model.summary()
```

```python
# Bottleneck 구조 Inception Module 구현  chatGPT

import tensorflow as tf

from tensorflow.keras import Model
from tensorflow.keras.layers import (
    Input,
    Dense,
    Conv2D,
    BatchNormalization,
    ReLU,
    MaxPooling2D,
    GlobalAveragePooling2D,
    Dropout,
    concatenate
)

# Conv2D → Batch Normalization → ReLU
def conv_bn_relu(x, filters, kernel_size, name):
    x = Conv2D(
        filters=filters,
        kernel_size=kernel_size,
        strides=1,
        padding="same",
        use_bias=False,                # BN 사용 시 편향 생략 가능
        kernel_initializer="he_normal",
        name=f"{name}_conv"
    )(x)

    x = BatchNormalization(
        name=f"{name}_bn"
    )(x)

    x = ReLU(
        name=f"{name}_relu"
    )(x)

    return x


# Inception 3a Bottleneck Module
def inception_3a(x):

    # 경로 1: 1×1 Conv
    branch1 = conv_bn_relu(
        x,
        filters=64,
        kernel_size=1,
        name="branch1_1x1"
    )

    # 경로 2: 1×1 Bottleneck → 3×3 Conv
    branch2 = conv_bn_relu(
        x,
        filters=96,
        kernel_size=1,
        name="branch2_reduce"
    )

    branch2 = conv_bn_relu(
        branch2,
        filters=128,
        kernel_size=3,
        name="branch2_3x3"
    )

    # 경로 3: 1×1 Bottleneck → 5×5 Conv
    branch3 = conv_bn_relu(
        x,
        filters=16,
        kernel_size=1,
        name="branch3_reduce"
    )

    branch3 = conv_bn_relu(
        branch3,
        filters=32,
        kernel_size=5,
        name="branch3_5x5"
    )

    # 경로 4: 3×3 Max Pooling → 1×1 Conv
    branch4 = MaxPooling2D(
        pool_size=(3, 3),
        strides=1,
        padding="same",
        name="branch4_pool"
    )(x)

    branch4 = conv_bn_relu(
        branch4,
        filters=32,
        kernel_size=1,
        name="branch4_1x1"
    )

    # 네 경로의 특징 맵을 채널 방향으로 결합
    output = concatenate(
        [branch1, branch2, branch3, branch4],
        axis=-1,
        name="inception_3a_output"
    )

    return output

# 전체 분류 모델
def build_inception_model(
    input_shape=(32, 32, 3),
    num_classes=10
):
    model_input = Input(
        shape=input_shape,
        name="input_image"
    )

    # 초기 특징 추출: 3채널 → 192채널
    pre = conv_bn_relu(
        model_input,
        filters=192,
        kernel_size=1,
        name="preprocess"
    )

    # Inception 3a Module
    x = inception_3a(pre)

    # 특징 맵 전체를 채널별 평균값으로 변환
    x = GlobalAveragePooling2D(
        name="global_average_pooling"
    )(x)

    # 과적합 방지
    x = Dropout(
        rate=0.4,
        name="dropout"
    )(x)

    # 최종 분류층
    model_output = Dense(
        units=num_classes,
        activation="softmax",
        kernel_initializer="glorot_uniform",
        name="classifier"
    )(x)

    model = Model(
        inputs=model_input,
        outputs=model_output,
        name="Inception_3a_Classifier"
    )

    return model

# 모델 생성
model = build_inception_model(
    input_shape=(32, 32, 3),
    num_classes=10
)

# 모델 컴파일
model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()
```

```python
# Residual Block 구현  chatGPT

from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Conv2D, Activation, Add

# 1. 입력층
inputs = Input(shape=(28, 28, 3), name="input")

# 2. 초기 합성곱층: 채널 수를 3개에서 128개로 변환
x = Conv2D(
    filters=128,
    kernel_size=3,
    padding="same",
    activation="relu",
    name="initial_conv"
)(inputs)

# Residual Block의 입력을 Shortcut 경로에 저장
shortcut = x

# 3. 첫 번째 합성곱층: 128채널 → 64채널
x = Conv2D(
    filters=64,
    kernel_size=3,
    padding="same",
    activation="relu",
    name="residual_conv1"
)(x)

# 4. 두 번째 합성곱층: 64채널 → 128채널
# 덧셈 후 ReLU를 적용하므로 여기서는 activation을 사용하지 않음
x = Conv2D(
    filters=128,
    kernel_size=3,
    padding="same",
    activation=None,
    name="residual_conv2"
)(x)

# 5. 잔차 출력 F(x)와 입력 x를 요소별로 덧셈
x = Add(name="residual_add")([x, shortcut])

# 6. 최종 활성화 함수
outputs = Activation("relu", name="residual_relu")(x)

# 7. 모델 생성
model = Model(
    inputs=inputs,
    outputs=outputs,
    name="Residual_Block_Model"
)

# 8. 모델 구조 출력
model.summary()
```

```python
# CIFAR-10 이미지 분류를 위한 ResNet 구현 사례 143page

from tensorflow.keras import Model 
from tensorflow.keras.layers import * 

inputs = Input(shape=(32,32,3))  # cifar 데이터 셋 사용

conv = Conv2D(32,3, padding='same', activation='relu')
x = conv(inputs)
x = BatchNormalization()(x)

conv = Conv2D(32,3, padding='same', activation='relu')
y = conv(x)
y = BatchNormalization()(y)

conv = Conv2D(32, 3, padding='same')
y = conv(y)
y = BatchNormalization()(y)

x = add([x, y])
x = Activation('relu')(x)

conv = Conv2D(32, 3, padding='same', activation='relu')
y = conv(x)
y = BatchNormalization()(y)

conv = Conv2D(32, 3, padding='same')
y = conv(y)
y = BatchNormalization()(y)

x = add([x,y])
x = Activation('relu')(x)

x = AveragePooling2D(pool_size=8)(x)
y = Flatten()(x)

y = Dense(512, activation='relu')(y)
outputs = Dense(10, activation='softmax')(y)

model = Model(inputs=inputs, outputs=outputs)
model.summary()


```

```python
# 데이터 불러오기  cifar10 
from tensorflow.keras.datasets import cifar10 
(train_X, train_y), (test_X, test_y) = cifar10.load_data()

# 픽셀값 정규화
train_X = train_X.astype("float32") / 255.0
test_X = test_X.astype("float32") / 255.0 
print('train_X shape: ', train_X.shape) 
```

```python
# callback  145page 

from tensorflow.keras.callbacks import *

def lr_schedule(epoch):
    lr = 1e-3
    if epoch > 90:
        lr *= 0.5e-3
    elif epoch > 80 :
        lr *= 1e-3
    elif epoch > 60:
        lr *= 1e-2
    elif epoch > 40:
        lr *= 1e-1
    print('Learning rate:', lr)
    return lr

lr_scheduler = LearningRateScheduler(lr_schedule)

checkpoint = ModelCheckpoint(filepath='model-cifar.keras',
                             monitor='val_accuracy', 
                             save_best_only=True)

tensorBoard = TensorBoard(log_dir='./cifar_log', 
                          histogram_freq=1, embeddings_freq=1)

model.compile(loss='sparse_categorical_crossentropy', 
              optimizer='adam', metrics=['accuracy'])

model.fit(train_X, train_y, validation_data=(test_X, test_y), 
          epochs=50, batch_size=200, 
          callbacks=[lr_scheduler, checkpoint, tensorBoard])

```
