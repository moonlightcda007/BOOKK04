```python
# chatGPT 코드 (SIGMOID, Thanh, ReLU )

import numpy as np
import matplotlib.pyplot as plt

# 입력값 생성
x = np.arange(-10, 10, 0.1)

# Sigmoid 함수
sigmoid = 1 / (1 + np.exp(-x))

# Tanh 함수
tanh = np.tanh(x)

# ReLU 함수
relu = np.maximum(0, x)

# 그래프 출력
plt.figure(figsize=(10, 6))

plt.plot(x, sigmoid, label='Sigmoid')
plt.plot(x, tanh, label='Tanh')
plt.plot(x, relu, label='ReLU')

plt.title("Activation Functions")
plt.xlabel("Input")
plt.ylabel("Output")

plt.grid(True)
plt.legend()

plt.show()

```

```python
# chatGPT  Binary step  함수로 구현

import numpy as np
import matplotlib.pyplot as plt

def binary_step(x):
    return np.where(x >= 0, 1, 0)

x = np.arange(-10, 10, 0.1)

plt.plot(x, binary_step(x))
plt.title("Binary Step Function")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True)

plt.show()
```

```python
# chatGPT  binary step

import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-4, 4, 1000)

y = [1 if data >= 0 else 0 for data in x]

plt.figure(figsize=(8,5))

plt.plot(x, y, color='green', linewidth=3)

plt.axvline(0, color='gray', linestyle='--')

plt.title("Binary Step Function", fontsize=18)

plt.xlabel("Input (x)")
plt.ylabel("Output f(x)")

plt.text(-3.5, 0.7,
    r'$f(x)=\left\{\begin{array}{ll}0 & x<0 \\ 1 & x\geq0\end{array}\right.$',
    fontsize=14)

plt.grid(True)
plt.ylim(-0.2, 1.2)

plt.show()
```

```python
# tanh 함수 

x = np.linspace(start=-3.5, stop=3.5, num=50)
y = np.tanh(x)

plt.plot(x,y, c='green', lw=8)
plt.axvline(0, c='grey')
plt.axhline(0, c='grey') 
plt.title(r'$tanh$', fontsize=20)
plt.text(-3.2, 0.75, r'$f(x)=\frac{2}{1+e^{-2x}}-1$', fontsize=15)
plt.show()
```

```python
# softmax 함수 

x = np.linspace(start=0, stop=10, num=10)
def softmax(input):
    return np.exp(input)/(np.sum(np.exp(input)))
y = softmax(x)

plt.plot(x,y, c='green', lw=8)
plt.ylim(-0.1, 1)
plt.axvline(0, c='grey')
plt.axhline(0, c='grey') 
plt.title(r'$softmax$', fontsize=20)
plt.text(2, 0.8, r'$f(x)=\frac{e^{x_i}}{\sum_{j=1}^Je^x_i}\/\/ \/\/for\/ i=1,\/\/...,\/\/j$', fontsize=15)
plt.show()
```

```python
# ReLU 함수 

x = np.linspace(start=-6, stop=6, num=50)
y = [data if data >= 0 else 0 for data in x]

plt.plot(x,y, c='green', lw=8)
plt.axvline(0, c='grey')
plt.axhline(0, c='grey') 
plt.title(r'$ReLU$', fontsize=20)
plt.text(-5, 5, r'$f(x)=\{  \genfrac{}{}{0}{}{0  \/\/  for\/\/ x<0}{1 \/ \/ for \/\/ x \geq 0}$', fontsize=15)
plt.show()
```

```python
# LeakyReLu & PReLU 함수 

x = np.linspace(start=-6, stop=6, num=50)
y = [data if data >= 0 else 0.1*data for data in x]

plt.plot(x,y, c='green', lw=8)
plt.axvline(0, c='grey')
plt.axhline(0, c='grey') 
plt.title(r'$LeakyReL(\alpha=0.1)$', fontsize=20)
plt.text(-5, 5, r'$f( \alpha, x)=\{  \genfrac{}{}{0}{}{\alpha x \/\/  for\/\/ x<0}{x \/\/ for \/\/ x \geq 0}$', fontsize=15)
plt.show()
```

```python
# ELU 함수 

x = np.linspace(start=-6, stop=6, num=50)
y = [data if data >= 0 else 1*(np.e**data-1) for data in x]

plt.plot(x,y, c='green', lw=8)
plt.axvline(0, c='grey')
plt.axhline(0, c='grey') 
plt.title(r'$ELU(\alpha=0.1)$', fontsize=20)
plt.text(-5, 5, r'$f( \alpha, x)=\{  \genfrac{}{}{0}{}{\alpha (e^x-1) \/\/  for\/\/ x<0}{x \/\/ for \/\/ x \geq 0}$', fontsize=15)
plt.show()
```

```python
# Gaussian 함수 

x = np.linspace(start=-4, stop=4, num=50)
y = np.exp(-(x**2))

plt.plot(x,y, c='green', lw=8)
plt.axvline(0, c='grey')
plt.axhline(0, c='grey') 
plt.title(r'$Gaussian$', fontsize=20)
plt.text(-4, 0.8, r'$f(x)=e^{-x^2}$', fontsize=15)
plt.show()
```

```python
# chatGPT 개선 코드

from tensorflow.keras.activations import softmax
from tensorflow.keras.layers import Input
from tensorflow.keras.layers import Dense
from tensorflow.keras.layers import ELU
from tensorflow.keras.layers import PReLU
from tensorflow.keras.models import Sequential

model = Sequential()

model.add(Input(shape=(4,)))

# 문자열 방식
model.add(Dense(50, activation='sigmoid'))

# 클래스 방식
model.add(Dense(50))
model.add(PReLU())

# 클래스 방식
model.add(Dense(30))
model.add(ELU(alpha=1))

# 함수 방식
model.add(Dense(3, activation=softmax))

model.summary() 
```

```python
# page 67 

from tensorflow.keras import Sequential
from tensorflow.keras.layers import Input, Dense, Dropout,BatchNormalization
from tensorflow.keras.initializers import RandomNormal

model = Sequential([
    Input(shape=(11,)),
    Dense(100, activation='sigmoid', kernel_initializer='he_normal'),
    Dropout(0.2),
    BatchNormalization(momentum=0.9, epsilon=1e-3),
    Dense(200, activation='relu', kernel_initializer='glorot_uniform'),
    Dropout(0.4),
    Dense(50, activation='tanh', kernel_initializer=RandomNormal(0, 0.1)),
    Dropout(0.1),
    Dense(6, activation='softmax')
])
model.summary()
# print(model.to_json)
```
